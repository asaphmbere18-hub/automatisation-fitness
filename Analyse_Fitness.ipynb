# PARTIE 2 : NETTOYAGE ET VALIDATION (5 points)
**Exercice 2.1 - Conversion des types de données**

Convertir toutes les colonnes de dates en format datetime :
* birth_date, registration_date dans members
* start_date, end_date dans subscriptions
* check_in_time, check_out_time dans visits
* opening_date dans centers

Attention : Certaines dates sont au format DD/MM/YYYY, d'autres en YYYY-MM-DD !

````python
# Conversion de toutes les dates en format datetime* birth_date, registration_date dans members start_date, end_date dans subscriptions check_in_time, check_out_time dans visits opening_date dans centers
# Attention : Certaines dates sont au format DD/MM/YYYY, d'autres en YYYY-MM-DD !
members['birth_date'] = pd.to_datetime(members['birth_date'], errors='coerce', dayfirst=True, format='mixed')
members['registration_date'] = pd.to_datetime(members['registration_date'], errors='coerce', dayfirst=True, format='mixed')
subsciptions['start_date'] = pd.to_datetime(subsciptions['start_date'], errors='coerce', dayfirst=True, format='mixed')
subsciptions['end_date'] = pd.to_datetime(subsciptions['end_date'], errors='coerce', dayfirst=True, format='mixed')
visits['check_in_time'] = pd.to_datetime(visits['check_in_time'], errors='coerce', dayfirst=True, format='mixed')
visits['check_out_time'] = pd.to_datetime(visits['check_out_time'], errors='coerce', dayfirst=True, format='mixed')
center['opening_date'] = pd.to_datetime(center['opening_date'], errors='coerce', dayfirst=True, format='mixed')
````

````python
print(members[['birth_date', 'registration_date']])
print(subsciptions[['start_date', 'end_date']])
print(visits[['check_in_time', 'check_out_time']])
print(center[['opening_date']])
````

**Exercice 2.2 - Détection et correction des dates incohérentes**

Identifier et marquer comme invalides (ou supprimer) :
* Les membres nés avant 1900 ou après 2024
* Les membres de moins de 16 ans (âge minimum légal)
* Les abonnements où end_date < start_date
* Les visites où check_out_time < check_in_time

````python
# Détection et correction des dates incohérentes
# identifier et marquer comme invalides (ou supprimer):
# Les membres nés avant 1900 ou après 2024
invalid_birth_dates = members[(members['birth_date'] < '1900-01-01') | (members['birth_date'] > '2024-12-31')]
print("\nDates de naissance invalides:\n", invalid_birth_dates)
# Les membres de moins de 16 ans (âge minimum légal)
current_date = visits['check_in_time'].max()
members['age'] = (current_date - members['birth_date']).dt.days // 365
invalid_age = members[members['age'] < 16]
print("\nMembres de moins de 16 ans:\n", invalid_age)
# Les abonnements où end_date < start_date
invalid_subscriptions = subsciptions[subsciptions['end_date'] < subsciptions['start_date']]
print("\nAbonnements avec end_date < start_date:\n", invalid_subscriptions)
# Les visites où check_out_time < check_in_time
invalid_visits = visits[visits['check_out_time'] < visits['check_in_time']]
print("\nVisites avec check_out_time < check_in_time:\n", invalid_visits)
# Suppression des lignes incohérentes détectées
subsciptions = subsciptions.drop(invalid_subscriptions.index)
visits = visits.drop(invalid_visits.index)
print("\nAbonnements restants :", len(subsciptions), "| Visites restantes :", len(visits))
````

# Exercice 2.4 - Nettoyage des prix
* Détecter les monthly_price négatifs ou supérieurs à 200€
* Détecter les incohérences : plan_name = "Premium" mais monthly_price < 30€

````python
# Détecter les monthly_price négatifs ou supérieurs à 200€
invalid_prices = subsciptions[(subsciptions['monthly_price'] < 0) | (subsciptions['monthly_price'] > 200)]
print("\nAbonnements avec monthly_price invalide:\n", invalid_prices)

# Détecter les incohérences : plan_name = "Premium" mais monthly_price < 30€
inconsistent_premium = subsciptions[(subsciptions['plan_name'] == 'Premium') & (subsciptions['monthly_price'] < 30)]
print("\nAbonnements Premium avec monthly_price < 30€:\n", inconsistent_premium)
# Un prix négatif est impossible : on le neutralise pour ne pas fausser total_paid
subsciptions.loc[subsciptions['monthly_price'] < 0, 'monthly_price'] = np.nan
````

# PARTIE 3 : ENRICHISSEMENT (4 points)
**Exercice 3.1 - Calcul de l'âge des membres**
* Créer une colonne age calculée à partir de birth_date
* Créer une colonne age_group :
 "16-25", "26-35", "36-50", "51-65", "65+"

````python
# Creer une colone age calculée à partir de birth_date
# Déja effectuée plus haut lors de la détection des dates incohérentes

#  Créer une colonne age_group : "16-25", "26-35", "36-50", "51-65", "65+"
bins = [15, 25, 35, 50, 65, 120]
labels = ['16-25', '26-35', '36-50', '51-65', '65+']
members['age_group'] = pd.cut(members['age'], bins=bins, labels=labels, right=True)
members.head(25)
````

# Exercice 3.2 - Extraction d'informations temporelles

À partir de check_in_time dans visits :
* Extraire : year, month, day_of_week (nom du jour), hour
* Créer is_weekend (booléen)
* Créer time_slot :
-  "Morning" (6h-12h)
- "Afternoon" (12h-18h)
- "Evening" (18h-22h)

````python
# Extraction : year, month, day_of_week (nom du jour), hour
visits['year'] = visits['check_in_time'].dt.year
visits['month'] = visits['check_in_time'].dt.month
visits['day_of_week'] = visits['check_in_time'].dt.day_name()
visits['hour'] = visits['check_in_time'].dt.hour

# Rajoutons minute pour plus de précision
visits['minute'] = visits['check_in_time'].dt.minute

# Créer is_weekend (booléen)
visits['is_weekend'] = visits['day_of_week'].isin(['Saturday', 'Sunday'])

# Créer time_slot
def assign_time_slot(hour):
    if 6 <= hour < 12:
        return 'Morning'
    elif 12 <= hour < 18:
        return 'Afternoon'
    elif 18 <= hour < 22:
        return 'Evening'
    else:
        return 'Night'
visits['time_slot'] = visits['hour'].apply(assign_time_slot)

visits.head(25)
````

# Exercice 3.3 - Calcul de la durée de visite
* Créer visit_duration_minutes = différence entre check-out et check-in
* Identifier les durées aberrantes (< 5 min ou > 300 min)
* Créer visit_duration_category :
* "Express" (< 30 min)
* "Normal" (30-90 min)
* "Extended" (> 90 min)

````python
# Création de visit_duration_minutes = différence entre check-out et check-in
visits['visit_duration_minutes'] = (visits['check_out_time'] - visits['check_in_time']).dt.total_seconds() / 60
print(visits[['check_in_time', 'check_out_time', 'visit_duration_minutes']].head())

# Identification des durées abérrantes (<5 minutes ou > 300 minutes)
durees_aberrantes = visits[(visits['visit_duration_minutes'] < 5) | (visits['visit_duration_minutes'] > 300)]
print("\nVisites avec des durées aberrantes:\n", durees_aberrantes)

# Création de visit_duration_category
def categorize_duration(minutes):
    if minutes < 30:
        return 'Express'
    elif 30 <= minutes <= 90:
        return 'Normal'
    else:
        return 'Extended'
visits['visit_duration_category'] = visits['visit_duration_minutes'].apply(categorize_duration)
print(visits[['visit_duration_minutes', 'visit_duration_category']].head())
````

# Exercice 3.4 - Calcul des métriques d'abonnement

À partir de subscriptions :
* Créer subscription_duration_days = durée de l'abonnement
* Créer is_active : True si end_date > aujourd'hui et payment_status = "paid"
* Créer total_paid = monthly_price × (subscription_duration_days / 30)

````python
# Création de subscription_duration_days =  duréé de l'abonnement
subsciptions['subscription_duration_days'] = (subsciptions['end_date'] - subsciptions['start_date']).dt.days
print(subsciptions[['start_date', 'end_date', 'subscription_duration_days']].head())

# Création de is_active : True si end_date > aujourd'hui et payment_statut ="paid"
subsciptions['is_active'] = (subsciptions['end_date'] > current_date) & (subsciptions['payment_status'] == 'paid')
print(subsciptions[['end_date', 'payment_status', 'is_active']].head(10))

# Création de total_paid = monthly_price * (subscription_duration_days / 30)
subsciptions['total_paid'] = subsciptions['monthly_price'] * (subsciptions['subscription_duration_days'] / 30)
print(subsciptions[['monthly_price', 'subscription_duration_days', 'total_paid']].head())
````

# PARTIE 4 : FUSION DES DONNÉES (2 points)
**Exercice 4.1 - Jointure members + subscriptions**
* Joindre members avec subscriptions sur member_id
* Type de jointure : left join (garder tous les membres)
* Résultat attendu : Un membre peut avoir plusieurs abonnements successifs

````python
# Jointures de members avec subscriptions sur memeber_id avec pour type de jointure left join
members_subscriptions = pd.merge(members, subsciptions, on='member_id', how='left', suffixes=('_member', '_subscription'))
print(members_subscriptions.head())
````

# Exercice 4.2 - Jointure avec visits
* Joindre le résultat précédent avec visits
* Identifier les visits orphelines (member_id absent de members.csv)
* Supprimer ces visites invalide

````python
# Identification des visits orphelines (member_id absent de members.csv)
visite_orpheline = visits[~visits['member_id'].isin(members['member_id'])]
print("\nVisites orphelines (member_id absent de members.csv):\n", visite_orpheline)

# Suppression des visits orphelines du DataFrame final
visite_clean = visits[visits['member_id'].isin(members['member_id'])]
# Joindre le résultat avec visits
members_subscriptions_visits = pd.merge(members_subscriptions, visite_clean, on='member_id', how='left', suffixes=('', '_visit'))
# Un membre avec plusieurs abonnements duplique chacune de ses visites.
# Pour chaque visite, on garde l'abonnement actif à la date de la visite s'il existe.
members_subscriptions_visits['abo_actif'] = (
    (members_subscriptions_visits['check_in_time'] >= members_subscriptions_visits['start_date']) &
    (members_subscriptions_visits['check_in_time'] <= members_subscriptions_visits['end_date'])
)
members_subscriptions_visits = members_subscriptions_visits.sort_values('abo_actif', ascending=False)
members_subscriptions_visits = members_subscriptions_visits.drop_duplicates(subset=['member_id', 'visit_id'])
# Contrôle : chaque visite ne doit apparaître qu'une seule fois
assert not members_subscriptions_visits['visit_id'].dropna().duplicated().any()
print(members_subscriptions_visits.shape)
````

**Exercice 4.3 - Jointure avec centers**
* Ajouter les informations du centre visité
* Vérifier que tous les center_id de visits existent dans centers

````python
# Ajout des informations du centre visité
df = pd.merge(members_subscriptions_visits, center, on='center_id', how='left')
print(df.shape)

# Vérifions que tous les center_id de visit existent dans center
missing_centers = visits[~visits['center_id'].isin(center['center_id'])]
print("\nVisites avec center_id absent de center.csv:\n", missing_centers)
if len(missing_centers) == 0:
    print("Tous les center_id de visits existent dans centers")
else:
    print("Certains center_id de visits sont absents de centers")
````

# PARTIE 5 : AGRÉGATIONS ET INDICATEURS (4 points)
**Exercice 5.1 - Calcul des KPIs par membre**

Pour chaque member_id, calculer :
* total_visits = nombre total de visites
* avg_visit_duration = durée moyenne de visite
* favorite_center = centre le plus fréquenté
* last_visit_date = date de dernière visite
* days_since_last_visit = nombre de jours depuis la dernière visite

````python
KPI_member = df.groupby('member_id').agg(
    total_visits = ('visit_id', 'count'),
    avg_visit_duration = ('visit_duration_minutes','mean'),
    last_visit_date = ('check_in_time', 'max')
)

# Calcul du centre favori
KPI_member['favorite_center'] = df.groupby('member_id')['center_name'].agg(lambda x: x.mode().iloc[0] if not x.mode().empty else None)

# calcul du nombre de jours depuis la dernière visite
KPI_member['days_since_last_visit'] = (current_date - KPI_member['last_visit_date']).dt.days

print(KPI_member.head())
````

# Exercice 5.2 - Calcul des KPIs par centre

Pour chaque center_id, calculer :
* total_unique_members = nombre de membres différents reçus
* avg_daily_visits = moyenne de visites par jour
* peak_hour = heure de pointe (heure avec le plus de visites)
* occupancy_rate = nombre de visites / capacité théorique

````python
centre_perfomants = df.groupby('center_id').agg(
    total_unique_members = ('member_id', 'nunique'),
    total_visites = ('visit_id', 'count'),
    capacity = ('surface_m2', 'first')
).reset_index()

# Heure de pointe
heure_pointe = df.groupby('center_id')['hour'].agg(lambda x: x.mode()[0])
centre_perfomants['peak_hour'] = centre_perfomants['center_id'].map(heure_pointe)
# # Moyenne de visites par jour
nb_jours = df['check_in_time'].dt.date.nunique()
centre_perfomants['avg_daily_visits'] = centre_perfomants['total_visites'] / nb_jours

# Taux d'occupation moyen
centre_perfomants['occupancy_rate'] = centre_perfomants['total_visites'] / centre_perfomants['capacity']
print(centre_perfomants.head())
````
